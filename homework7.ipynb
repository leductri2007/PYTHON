Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu
mỗi phần tử x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp

```python
a=list(map(int,input().split()))
d={}
so_can_xoa=0
mang_dep=True
y=0
for x in a:
    d[x]=d.get(x,0)+1
for key, val in d.items():
    if key==val:
        continue
    elif key<val:
        mang_dep=False
        so_can_xoa=val-key
        y+=so_can_xoa
    else:
        mang_dep=False
        so_can_xoa=val
        y+=so_can_xoa
if mang_dep:
    print(True)
else:
    print(y)
```

Tìm số có đúng k chữ số lớn nhất có thể
trong một số có n chữ số (coi đầu vào là
một số rất lớn - str). Không được hoán
đổi vị trí các chữ số


```python
s = input().strip()
k = int(input())

n = len(s)
ket_qua = ""
vi_tri_bat_dau = 0

for i in range(k):
    # Giới hạn vị trí cuối cùng được phép chọn để đảm bảo phía sau còn đủ ký tự
    vi_tri_ket_thuc = n - (k - 1 - i)

    # Cắt lấy đoạn chuỗi hợp lệ được phép tìm kiếm
    doan_xet = s[vi_tri_bat_dau:vi_tri_ket_thuc]

    # Tìm chữ số lớn nhất trong đoạn này
    max_val = max(doan_xet)

    # Nối chữ số lớn nhất vừa tìm được vào kết quả
    ket_qua += max_val

    # Cập nhật lại vị trí bắt đầu cho lượt tìm kiếm tiếp theo
    # Phải bắt đầu từ ngay sau vị trí của chữ số max_val vừa tìm được
    vi_tri_bat_dau += doan_xet.index(max_val) + 1

print(ket_qua)
```

̀i toán Tìm dãy các hợp số dài nhất liên
tiếp không chứa số nguyên nào <= N (N
là số nguyên dương cho trước). Nếu có
nhiều dãy cùng độ dài thì lấy dãy nhỏ
nhất (gặp đầu tiên từ trái qua)

```python
import math
def nt(n):
    if n<2:
        return False
    for i in range(2,int(math.sqrt(n))+1):
        if n%i==0:
            return False
    return True
n=int(input())
max_len=0
dem_hien_tai=0
bat_dau=0
for i in range(2,n+2):
    if i<=n and not nt(i):
        dem_hien_tai+=1
    else:
        if dem_hien_tai>max_len:
            max_len=dem_hien_tai
            bat_dau=i-dem_hien_tai
        dem_hien_tai=0
if max_len>0:
    print(*list(range(bat_dau,bat_dau+max_len)))
```
