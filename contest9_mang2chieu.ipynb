Cho ma trận n x m, tìm số nhỏ nhất và lớn nhất trong ma trận và liệt kê các vị trí xuất hiện của các phần tử này trong ma trận

```python
n,m=map(int,input().split())
a=[]
for _ in range(n):
    b=list(map(int,input().split()))
    a.append(b)
c=[]
for x in a:
    c.extend(x)
for i in range(n):
    for j in range(m):
        if a[i][j]==min(c):
            print(min(c))
            print(i+1,j+1)
        elif a[i][j]==max(c):
            print(max(c))
            print(i+1,j+1)
```

Đếm số ptu là snt trên đường chéo chính và phụ, mỗi phần tử chỉ đếm 1 lần (n x n)

```python
import math
def nt(n):
    if n<2:
        return False
    for i in range(2,int(math.sqrt(n))+1):
        if n%i==0:
            return False
    return True

n=int(input())
a=[]
for i in range(n):
    b=list(map(int,input().split()))
    a.append(b)

s=set()
for i in range(n):
    if nt(a[i][i]):
        s.add(a[i][i])
    elif nt(a[i][n-i-1]):
        s.add(a[i][n-1-i]) #lưu ý phần tử ở giữa bị đếm 2 lần nếu n lẻ
print(len(s))
```

Sắp xếp tăng dần theo cột

```python
n,m=map(int,input().split())
a=[]
for _ in range(n):
    b=list(map(int,input().split()))
    a.append(b)
for j in range(m): #duyệt theo cột phải for j trước
    cot=[]
    for i in range(n):
        cot.append(a[i][j])
    cot.sort()
    for i in range(n): #cột j đang thay đổi (đang trong for j)
        a[i][j]=cot[i]
for x in a:
    print(*x)
    
```

Cho ma trận A nxm, ma trận B mxp, in ra ma trận tích của ma trận A*B

```python
n,m,p=map(int,input().split())
A=[]
for i in range(n):
    z=list(map(int,input().split()))
    A.append(z)
B=[]
for j in range(m):
    v=list(map(int,input().split()))
    B.append(v)
C=[]
for i in range(n):
    hang=[]
    for k in range(p):
        tong=0
        for j in range(m): #lấy từng ptu trên hàng của A nhân với từng ptu tương ứng trên cột của B rồi cộng lại
            tong+=A[i][j]*B[j][k]
        hang.append(tong)
    C.append(hang) # C có n hàng p cột
for x in C:
    print(*x)
```

Xây dựng ma trận xoắn ốc cỡ n

```python
n=int(input())
a=[]
for _ in range(n):
    a.append([0]*n)
dem=0
h1,h2,c1,c2=0,n-1,0,n-1
while dem<n**2:
    for i in range(c1,c2+1): # Xây hàng trên cùng c1->c2
        a[h1][i]=dem
        dem+=1
    h1+=1
    for i in range(h1,h2+1): # Xây cột ngoài cùng h1->h2
        a[i][c2]=dem
        dem+=1
    c2-=1
    for i in range(c2,c1-1,-1): # Xây hàng dưới cùng c2->c1
        a[h2][i]=dem
        dem+=1
    h2-=1
    for i in range(h2,h1-1,-1): # Xây cột trong cùng h2->h1
        a[i][c1]=dem
        dem+=1
    c1+=1
for x in a:
    print(*x) 
```

Xây dựng ma trận xoắn ốc cỡ n toàn số fibonacci


```python
F=[0]*100
F[0]=0
F[1]=1
for i in range(2,100):
    F[i]=F[i-1]+F[i-2]

n=int(input())
a=[]
for _ in range(n):
    a.append([0]*n)
idex=0
h1,h2,c1,c2=0,n-1,0,n-1
while idex<n**2:
    for i in range(c1,c2+1): # Xây hàng trên cùng c1->c2
        a[h1][i]=F[idex]
        idex+=1
    h1+=1
    for i in range(h1,h2+1): # Xây cột ngoài cùng h1->h2
        a[i][c2]=F[idex]
        idex+=1
    c2-=1
    for i in range(c2,c1-1,-1): # Xây hàng dưới cùng c2->c1
        a[h2][i]=F[idex]
        idex+=1
    h2-=1
    for i in range(h2,h1-1,-1): # Xây cột trong cùng h2->h1
        a[i][c1]=F[idex]
        idex+=1
    c1+=1
for x in a:
    print(*x) 
```

Ma trận xoắn ốc giảm dần

Yêu cầu
Cho số nguyên dương n. Hãy tạo ma trận vuông kích thước n × n theo dạng xoắn ốc, bắt đầu từ góc trên bên trái, theo chiều kim đồng hồ, với các giá trị giảm dần từ n² về 1.

```python
n=int(input())
a=[]
for _ in range(n):
    a.append([0]*n)
dem=n**2
h1,h2,c1,c2=0,n-1,0,n-1
while dem>0:
    for i in range(c1,c2+1):
        a[h1][i]=dem
        dem-=1
    h1+=1
    for i in range(h1,h2+1):
        a[i][c2]=dem
        dem-=1
    c2-=1
    for i in range(c2,c1-1,-1):
        a[h2][i]=dem
        dem-=1
    h2-=1
    for i in range(h2,h1-1,-1):
        a[i][c1]=dem
        dem-=1
    c1+=1
for x in a:
    print(*x)

 
```

Bài 4: Xoắn ốc bung ra - Từ trong ra ngoài (Mức độ: Khó)Yêu cầu: Cho một số nguyên lẻ n (ví dụ: 3, 5, 7...). Hãy điền các số từ 1 đến n^2 vào ma trận cấp n x n theo quy luật: Số 1 nằm ở chính giữa ma trận, các số tiếp theo sẽ chạy theo đường xoắn ốc "bung" dần ra ngoài theo chiều kim đồng hồ.

```python
n=int(input())
a=[]
for _ in range(n):
    a.append([0]*n)
h1,h2,c1,c2=0,n-1,0,n-1
dem=n**2
while dem>=1:
    for i in range(c2,c1-1,-1):
        a[h1][i]=dem
        dem-=1
    h1+=1
    for i in range(h1,h2+1):
        a[i][c1]=dem
        dem-=1
    c1+=1
    for i in range(c1,c2+1):
        a[h2][i]=dem
        dem-=1
    h2-=1
    for i in range(h2,h1-1,-1):
        a[i][c2]=dem
        dem-=1
    c2-=1
for x in a:
    print(*x)
```
