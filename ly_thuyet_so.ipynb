Số nguyên tố

```python
import math
def prime(n):
    if n<2:
        return False
    
    for i in range(2,int(math.sqrt(n)+1)):
        if n%i==0:
            return False
    return True
```

Số chính phương

```python
import math
def scp(n):
    can=int(math.sqrt(n))
    if can*can==n:
        return True
    return False

a=int(input())
if scp(a):
    print(True)
else:
    print(False)
```

Viết chương trình nhập vào một số nguyên dương b. Tìm và in ra tất cả các "số hoàn hảo" nằm trong khoảng từ 1 đến n. Số hoàn hảo là số có tổng các ước số thực sự (không tính chính nó) bằng chính nó.

```python
n=int(input())
import math
for i in range(1,n+1):
    tong=0
    for j in range(1,i):
        if i%j==0:
            tong=tong+j            
    if tong==i:
        print(i, end=" ")
```

Viết chương trình in ra n cặp số nguyên tố sinh đôi đầu tiên (hai số nguyên tố cách nhau đúng 2 đơn vị, ví dụ: 3 và 5, 5 và 7, 11 và 13).

```python
import math
def prime(n):
    if n<2:
        return False
    for i in range(2,int(math.sqrt(n)+1)):
        if n%i==0:
            return False
    return True

n=int(input())
dem=0
snt1=3
while dem<n:
    if prime(snt1) and prime(snt1+2):
        print(snt1,snt1+2,end=" ")
        dem+=1
    snt1+=2
```
