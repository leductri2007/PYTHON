Bài 1: Tính điểm trung bình từ file
Hãy khởi tạo một file có tên là grades.txt tại thư mục đang làm việc với cấu trúc sau:
Alice 85
Bob 90
Charlie 78
Diana 92
Viết chương trình thực hiện:
- Mở file và đọc nội dung file ra terminal
- Tính toán và in điểm trung bình của các học sinh ra terminal sau đó ghi vào cuối file với nội
dung: Average: <điểm trung bình>

```python
tong=0
dem=0
with open("grades.txt","r") as f:
    for line in f:
        print(line.strip())
        ten,diem=line.split()
        tong+=int(diem)
        dem+=1
avg=tong/dem
print(avg)
with open("grades.txt","a") as f:
    f.write(f"{avg}\n")

```

Bài 2: Đếm số lượng từ trong file
Khởi tạo file essay.txt với nội dụng: I am \n a student from UET.
Viết chương trình thực hiện:
- Đến và in ra tổng số từ có trong file trên.

```python

with open("essay.txt","r") as f:
    word=f.read()
    w=word.split()

tong_so_tu=len(w)
print(tong_so_tu)
```

Bài 3: Đảo ngược nội dung file
Hãy tạo một file có tên quotes.txt với nội dung: No pain no gain\nNo money no party.
Viết chương trình thực hiện đọc file và in ra nội dung bị đảo ngược

```python
with open("quotes.txt","w") as f:
    f.write("No pain no gain\nNo money no party.")
with open("quotes.txt","r") as f:
    print(f.read())
    
```
