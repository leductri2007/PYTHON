Xây dựng lớp Triangle(a, b, c) kế thừa từ Polygon và ghi đè perimeter() trả về a+b+c.
Input
Một dòng chứa 3 số nguyên a b c.
Output
In ra chu vi tam giác. 


```python
class Polygon: #đa giác
    def perimeter(self):
        pass
class Triangle(Polygon): #tam giác
    def __init__(self,a,b,c):
        self.a=a
        self.b=b
        self.c=c
    def perimeter(self):
        return self.a+self.b+self.c
a,b,c=map(int,input().split())
hinh=Triangle(a,b,c)
s=hinh.perimeter()
print(s)
```

W8A22 Yêu cầu
Xây dựng lớp Matrix2x2(a, b, c, d) biểu diễn ma trận:
[a b]
[c d]
Viết phương thức trace() trả về a + d và determinant() trả về ad - bc.
Input
Một dòng chứa 4 số nguyên a b c d.
Output
In ra trace và determinant trên cùng một dòng. 

```python
class Matrix2x2:
    def __init__(self,a,b,c,d):
        self.a=a
        self.b=b
        self.c=c
        self.d=d
    def trace(self):
        return self.a+self.d
    def determinant(self):
        return self.a*self.d-self.b*self.c
a,b,c,d=map(int,input().split())
tong=Matrix2x2(a,b,c,d)
x=tong.trace()
y=tong.determinant()
print(x,y)
```

W8A27 Yêu cầu
Xây dựng lớp Product(name, price).
Xây dựng lớp DiscountProduct(name, price, percent) kế thừa từ Product, ghi đè phương thức
final_price() để trả về giá sau giảm.
Công thức: price * (100 - percent) / 100
Input
name price percent
Output
In ra giá cuối cùng với 2 chữ số thập phân. 

```python
class Product:
    def __init__(self,name,price):
        self.name=name
        self.price=price
    def final_price(self):
        pass
class DiscountProduct(Product):
    def __init__(self, name, price,percent):
        super().__init__(name, price)
        self.percent=percent
    def final_price(self):
        return self.price * (100 - self.percent) / 100
name=input()
price,percent=map(int,input().split())
a=DiscountProduct(name,price,percent)
b=a.final_price()
print(f"{b:.2f}")
```

W8A28 Yêu cầu
Xây dựng lớp Counter có thuộc tính count ban đầu bằng 0 và các phương thức inc(), dec(),
value().
Sau đó thực hiện dãy thao tác trên đối tượng.
Input
Dòng 1: n
Dòng 2: n thao tác, mỗi thao tác là INC hoặc DEC, cách nhau bởi dấu cách.
Output
Giá trị cuối cùng của count. 

```python
class Counter:
    def __init__(self):
        self.count = 0
    def inc(self):
        self.count += 1
    def dec(self):
        self.count -= 1
    def value(self):
        return self.count
n = int(input())
ops = input().split()
c = Counter()
for op in ops:
    if op == "INC":
        c.inc()
    elif op == "DEC":
        c.dec()
print(c.value())
```

W8A29 Yêu cầu
Xây dựng lớp Time(h, m, s) với phương thức to_seconds() đổi thời gian sang tổng số giây.
Input
h m s
Output
Tổng số giây

```python
class Time:
    def __init__(self,h,m,s):
        self.h=h
        self.m=m
        self.s=s
    def to_seconds(self):
        tong_giay=self.h*3600+self.m*60+self.s
        return tong_giay
h,m,s=map(int,input().split())
a=Time(h,m,s)
print(a.to_seconds())
```

W8A30 Yêu cầu
Xây dựng lớp Stack sử dụng danh sách để cài đặt ngăn xếp với các phương thức push(x),
pop(), top().
Dữ liệu đảm bảo mọi lệnh pop và top đều hợp lệ.
Input
Dòng 1: n
n dòng tiếp theo là một trong ba lệnh:
PUSH x
POP
TOP
Output
Mỗi lệnh TOP in ra giá trị ở đỉnh ngăn xếp trên một dòng. 

```python
class Stack:
    def __init__(self):
        self.s = []
    def push(self, x):
        self.s.append(x)
    def pop(self):
        self.s.pop()
    def top(self):
        return self.s[-1]

n = int(input())
st = Stack()
for _ in range(n):
    cmd = input().split()
    if cmd[0] == "PUSH":
        st.push(cmd[1]) #vì nhập vào PUSH "x" (thêm vào x)
    elif cmd[0] == "POP":
        st.pop()
    elif cmd[0] == "TOP":
        print(st.top())
```
