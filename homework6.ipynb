Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất trong list


```python
def gtln(a):
    ln=0
    val=0
    for i in range(len(a)):
        if ln<a[i]:
            ln=a[i]
            val=i
    return val
a=list(map(int,input().split()))
print(gtln(a))
```

trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải

đường chéo phụ

```python
n=int(input())
a=[]
for i in range(n):
    b=list(map(int,input().split()))
    a.append(b)

for i in range(n):
    for j in range(n):
        if i==j:
            print(a[i][j],end=" ")

print()
for x in range(n):
    for y in range(n):
        if x+y==n-1:
            print(a[x][y],end=" ")
```

Nhâ ̣p vào mảng n hàng, m cột số nguyên và số nguyên k (n, m, k nhập trên ̣
cùng một dòng, cách nhau dấu cách). Tính và in ra tổng các phần tử trên cột
k (hàng k)

W6A20
Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về
-1

```python
def check(a,k):
    for i in range(len(a)):
        if a[i]==k:
            return i
    else:
        return -1
a=list(map(int,input().split()))
k=int(input())
print(check(a,k))
```

Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện.
Bài này yêu cầu:

Nhập 1 dòng gồm các chuỗi (cách nhau bởi dấu cách).
Với mỗi chuỗi, lưu lại tất cả vị trí xuất hiện (0-based).
Giá trị trong dictionary là tuple

```python
a=input().split()
b=list(a)
d={}
for i in range(len(b)):
    if b[i] in d:
        d[b[i]].append(i)
    else:
        d[b[i]]=[i]
for key, value in d.items():
    d[key]=tuple(value)
print(d)    
#***
```

W6A23 VIết hàm xóa đi các số <x có trong mảng

```python
def xoa(a,x):
    b=[]
    for k in a:
        if k>=x:
            b.append(k)
    return b
a=list(map(int,input().split()))
x=int(input())
print(xoa(a,x))
```

W6A24
Viêết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9


```python
def day(a):
    b=[]
    c=[]
    for x in a:
        if x==0:
            b.append(x)
        else:
            c.append(x)
    
    return b+c
a=list(map(int,input().split()))
print(day(a))
```
