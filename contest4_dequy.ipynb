Tính tổng S(n)=1+2+3+...+n bằng đệ quy

```python
def tong(n):
    if n==0:
        return 0
    return n+tong(n-1)
print(tong(4))
```

Tính S(n)=1^2+2^2+3^2+4^2+...+n^2

```python
def tong(n):
    if n==0:
        return 0
    return n**2+tong(n-1)
print(tong(3))
```

Tính S(n)=-1+2-3+4-5+6-...+(-1)^n*n

```python
def tong(n):
    if n==0:
        return 0
    return (-1)**n*n+tong(n-1)
print(tong(2))
```

Tính n!

```python
def giai_thua(n):
    if n==1:
        return 1
    return n*giai_thua(n-1)
print(giai_thua(3))
```

Tìm số fibo thứ n sử dụng đệ quy

```python
def fibo(n):
    if n==1 or n==2:
        return n
    return fibo(n-1)+fibo(n-2)
print(fibo(5))
```

```python
def tong(n):
    if n==1:
        return 1
    return 1/n+tong(n-1)
print(tong(3))
```

```python
def np(n):
    if n!=0:
        return f"{np(n//2)}{n%2}"
    return ""
print(np(1))
    
```

Tính tổng chữ số

```python
def tong(n):
    if n<10:
        return n
    return n%10 + tong(n//10)
print(tong(223))
```

Đếm số lượng chữ số

```python
def cnt(n):
    if n<10:
        return 1
    return 1+cnt(n//10)
print(cnt(1000000000))
```

Tìm chữ số đầu tiên

```python
def first_num(n):
    if n<10:
        return n
    return first_num(n//10)
print(first_num(90023))
```

Tìm chữ số lớn nhất và nhỏ nhất của n theo đệ quy

```python
def max_num(n):
    if n<10:
        return n
    return max(n%10,max_num(n//10))
print(max_num(879))
```

Cho số nguyên dương n, in ra các chữ số của n theo thứ tự từ trái qua phải và từ phải qua trái

```python
def trai_qua_phai(n):
    if n<10:
        return n
    return f"{trai_qua_phai(n//10)} {n%10}"
print(trai_qua_phai(1234))
# tương tự
```

Tính tổng các chữ số chẵn, lẻ của n

```python
def tong_chan(n):
    if n<10:
        if n%2==0:
            return n
        else:
            return 0
    else:
        if n%2==0:
            return n%10+tong_chan(n//10)
        else:
            return tong_chan(n//10)

print(tong_chan(124352))
```

Kiểm tra tất cả các chữ số của n có phải đều là số chẵn hay không

```python
def check(n):
    if n<10:
        if n%2!=0:
            return False
        return True
    else:
        if n%2!=0:
            return False
        return check(n//10)
    
print(check(422412))
```

Thức hiện 3 thao tác: lấy n chia 2 nếu n chia hết cho 2, lấy n chia 3 nếu n chia hết cho 3, nếu không giảm n đi 1 đơn vị. Đếm số thao tác ít nhất để biến n về 1

```python
def bien_doi(n):
    tt1,tt2,tt3=10**9,10**9,10**9
    if n==1:
        return 0
    else:
        if n%2==0:
            tt1=1+bien_doi(n//2)
        elif n%3==0:
            tt2=1+bien_doi(n//3)
        else:
            tt3=1+bien_doi(n-1)
    return min(tt1,tt2,tt3)
print(bien_doi(9))
```

Kiểm tra tính đối xứng của mảng bằng đệ quy

```python
def check(a,l,r):

    if l>=r:
        return True
    if a[l]!=a[r]:
        return False
    else: 
        return check(a,l+1,r-1)
    
a=list(map(int,input().split()))
print(check(a,0,len(a)-1))
```

Kiểm tra mảng toàn chẵn

```python
def check(a,n):
    if a[0]%2==0:
        return True
    else:
        if a[n]%2!=0:
            return False
        return check(a,n-1)
a=list(map(int,input().split()))
print(check(a,len(a)-1))
```

Kiểm tra mảng tăng dần

```python
def check(a,n):
    if n==0:
        return True
    if a[n]<a[n-1]:
        return False
    return check(a,n-1)
a=list(map(int,input().split()))
print(check(a,len(a)-1))
```
