Sắp xếp trộn: Cho 2 list B và C đã sắp xếp tăng dần, trộn B và C thành list A tăng dần

```python
b=list(map(int,input().split()))
c=list(map(int,input().split()))
i,j=0,0

while i<len(b) and j<len(c):
    if b[i]<=c[j]:
        print(b[i],end=" ")
        i+=1
    else:
        print(c[j],end=" ")
        j+=1

# sau vòng lặp sẽ có trường hợp i hoặc j kết thúc trước, in ra nốt cái còn lại
while i<len(b):
     print(b[i],end=" ")
     i+=1
while j<len(c):
    print(c[j],end=" ")
    j+=1

```

Cho list a là list chiều cao của các bạn nam, list b là chiều cao của các bạn nữ, ghép đôi nếu bạn nam cao hơn bạn nữ, mỗi bạn chỉ có 1 đôi. Đếm số lượng đôi nhiều nhất có thể có

```python
a=list(map(int,input().split()))
b=list(map(int,input().split()))
a.sort()
b.sort()
i,j=0,0
dem=0

while i<len(a) and j<len(b):
     if a[i]>b[j]:
         dem+=1
         i+=1
         j+=1
     else:
         i+=1
print(dem) 
```

Cho n viên gạch đánh số từ 1 đến n, các viên gạch có độ cứng a1, a2,...an. Viên gạch có độ cứng x tức là có thể chồng lên trên viên gạch đó tối đa x viên gạch khác. Hỏi chồng gạch cao nhất có thể xếp là bao nhiêu

```python
a=list(map(int,input().split()))
a.sort(reverse=True)
dem=1
p=a[0]
for i in range(1,len(a)):
    if p<=0:
        break
    else:
        dem+=1
        p=min(p-1,a[i])
```

Cho list a có n phần tử và số nguyên dương k, đếm số lượng cặp số a[i]+a[j]=k   ( i!=j ) 

```python
# Số lớn nên dùng binary search
def b_first(a,l,r,x):
    kq=-1
    while l<=r:
        m=(l+r)//2
        if a[m]==x:
            kq=m
            r=m-1
        elif a[m]<x:
            l=m+1
        else:
            r=m-1
    return kq

def b_last(a,l,r,x):
    kq1=-1
    while l<=r:
        m=(l+r)//2
        if a[m]==x:
            kq1=m
            l=m+1
        elif a[m]<x:
            l=m+1
        else:
            r=m-1
    return kq1
a=list(map(int,input().split()))
a.sort()
k=int(input())
cnt=0
for i in range(len(a)):
    p,q=b_first(a,i+1,len(a)-1,k-a[i]),b_last(a,i+1,len(a)-1,k-a[i])
    if p!=-1:
        cnt+=q-p+1
print(cnt)
```

Cho list a, gọi l và r là giá trị min max của a, tìm số phần tử cần thêm vào để list có đầy đủ các số trong khoảng [l,r]

```python
a=list(map(int,input().split()))
l=min(a)
r=max(a)
dem=0
for i in range(l,r+1):
    if i not in a:
        dem+=1
print(dem)
```

Tìm cặp a[i]-a[j]=x, nếu tồn tại, print True

```python
def binary_search(a,l,r,x):
    while l<=r:
        m=(l+r)//2
        if a[m]==x:
            return True
        elif a[m]<x:
            l=m+1
        else:
            r=m-1
    return False
a=list(map(int,input().split()))
x=int(input())
for i in range(len(a)):
    if binary_search(a,0,len(a)-1,a[i]+x):
        print(True)
        break
else:
    print(False)
```
