Bài 4 Yêu cầu: Viết một hàm kiểm tra xem một chuỗi có phải là chuỗi đối xứng (palindrome) hay không, tức là đọc xuôi hay ngược đều giống
nhau.
Việc kiểm tra là chính xác tuyệt đối (phân biệt chữ hoa/thường và tính cả khoảng trắng). Trả về True hoặc False. Nếu đầu vào không phải
chuỗi, hãy chuyển nó thành chuỗi.

```python
def check(s):
    a=list(s)
    b=a.copy()
    b.reverse()
    if a==b:
        return True
    else: return False
s=input()
print(check(s))
```

Bài 5 Yêu cầu: Cho số nguyên n, hãy trả về tổng các chữ số của nó. Bỏ qua dấu âm (dùng giá trị tuyệt đối). 

```python
a=int(input())
n=abs(a)
tong=0
while n!=0:
    tong+=n%10
    n=n//10
print(tong)
```

Bài 6 Yêu cầu: Cho số nguyên không âm n, hãy tạo số lớn nhất và số nhỏ nhất có thể tạo được bằng cách sắp xếp lại các chữ số của n.
Trả về dưới dạng danh sách [max_value, min_value] (không dùng tuple).
Với số nhỏ nhất, không được để chữ số 0 đứng đầu (ví dụ: 2015 → 5210 và 1025).

```python
n=int(input())
a=str(n)
b=list(a)
c=b.copy()
c.sort()
b.sort(reverse=True)
max_val="".join(b)
if c[0]=="0":
    for i in range(len(c)):
        if c[i]!="0":
            c[0],c[i]=c[i],c[0]
            break


min_val="".join(c)
print([int(max_val),int(min_val)])
```
