Số Sphenic là số có thể phân tích duy nhất thành tích của 3 thừa số nguyên tố khác nhau. Cho số tự nhiên n, kiểm tra n có phải số Sphenic hay không

```python
import math
def snt(n):
    if n<2:
        return False
    if n>=2:
        for i in range(2,int(math.sqrt(n)+1)):
            if n%i==0:
                return False
        return True

def sphenic(a):
    dem=0
    tich=1
    for i in range(2,a//2):
        if snt(i) and a%i==0:
            tich*=i
            dem+=1
    if dem==3 and tich==a:
        return True
    return False

b=int(input())
if sphenic(b):
    print(True)
else:
    print(False)
```

```python
import math
def tong_so(n):
    tong=0
    while n>0:
        tong+=n%10
        n=n//10
    return tong

def tong_tsnt(a):
    tong1=tong_so(a)
    tong2=0
    k=a
    for i in range(2,int(math.sqrt(a)+1)): 
        while a%i==0:
            tong2+=tong_so(i)
            a=a//i
    if a==k:  #a là snt, không chia hết cho số nào ngoài 1 và chính nó nên chia xong vẫn =a
        return False
    if a>1:
        tong2+=tong_so(a)
    return tong1==tong2

c=int(input())
if tong_tsnt(c):
    print(True)
else:
    print(False)
```

Số thuần nguyên tố: số đó là snt, các chữ số là snt, tổng các chữ số là snt. Đếm xem trong đoạn a b có bao số thuần nguyên tố

```python
import math
def snt(n):
    if n<2:
        return False
    for i in range(2,int(math.sqrt(n))+1):
        if n%i==0:
            return False
    return True

def chu_so_snt(n):
    k=0
    while n>0:
        k=n%10
        if not snt(k):
            return False
        else:
            n=n//10
    return True

def tong_snt(n):
    tong=0
    while n>0:
        tong+=n%10
        n=n//10
    if snt(tong):
        return True
    return False

a,b=map(int,(input().split()))
dem=0
for i in range(a,b+1):
    if snt(i) and chu_so_snt(i) and tong_snt(i):
        dem+=1
print(dem)
```

Liệt kê các số nguyên tố nhỏ hơn hoặc bằng n và có chữ số hàng đơn vị lớn nhất(so với các chữ số của số đó), đếm xem có bao nhiêu số thỏa mãn

```python
import math
def snt(n):
    if n<2:
        return False
    for i in range(2,int(math.sqrt(n))+1):
        if n%i==0:
            return False
    return True

def check_dvi(n):
    dvi=n%10
    while n>0:
        n=n//10
        k=n%10
        if dvi<k:
            return False
    return True
        
a=int(input())
for i in range(2,a+1):
    if snt(i) and check_dvi(i):
        print(i, end=" ")

        
```
