Số Sphenic là số có thể phân tích duy nhất thành tích của 3 thừa số nguyên tố khác nhau. Cho số tự nhiên n, kiểm tra n có phải số Sphenic hay không

```python
import math
def snt(n):
    if n<2:
        return False
    if n>=2:
        for i in range(2,int(math.sqrt(n)+1)):
            if n%i==0:
                return False
        return True

def sphenic(a):
    dem=0
    tich=1
    for i in range(2,a//2):
        if snt(i) and a%i==0:
            tich*=i
            dem+=1
    if dem==3 and tich==a:
        return True
    return False

b=int(input())
if sphenic(b):
    print(True)
else:
    print(False)
```

Số thuần nguyên tố: số đó là snt, các chữ số là snt, tổng các chữ số là snt. Đếm xem trong đoạn a b có bao số thuần nguyên tố

```python
import math
def snt(n):
    if n<2:
        return False
    for i in range(2,int(math.sqrt(n))+1):
        if n%i==0:
            return False
    return True

def chu_so_snt(n):
    k=0
    while n>0:
        k=n%10
        if not snt(k):
            return False
        else:
            n=n//10
    return True

def tong_snt(n):
    tong=0
    while n>0:
        tong+=n%10
        n=n//10
    if snt(tong):
        return True
    return False

a,b=map(int,(input().split()))
dem=0
for i in range(a,b+1):
    if snt(i) and chu_so_snt(i) and tong_snt(i):
        dem+=1
print(dem)
```
