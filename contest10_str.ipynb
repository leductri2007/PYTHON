Cho 2 xâu s1 s1, in ra các ký tự xuất hiện ở cả 2 xâu theo từ điển, mỗi kí tự chỉ liệt kê 1 lần. Sau đó tiếp tục liệt kê các kí tự xuất hiện ở 1 trong 2 xâu theo thứ tự từ điển.

```python
s1=input()
s2=input()
s=set(s1)&set(s2)
ss=set(s1)|set(s2)
d=(list(s))
d.sort()
d1=(list(ss))
d1.sort()
print(d,end=' ')
print()
print(d1,end=" ")
```

Cho 2 xâu s1 s2, tìm các kí tự xuất hiện trong s1 mà k xuất hiện trong s2, các ký tự được in ra theo thứ tự từ điển và mỗi ký tự chỉ in 1 lần

```python
s1=input()
s2=input()
a=list(set(s1))
b=list(set(s2))
a.sort()
for x in a:
    if x not in b:
        print(x, end=" ")
```

Cho s là xâu kí tự gồm các từ và các dấu cách. Đầu tiên sắp xếp các từ trong xâu theo thứ tự từ điển tăng dần, sau đó sắp xếp theo thứ tự chiều dài tăng dần, trong trường hợp nhiều từ có cùng chiều dài thì từ nào có thứ tự từ điển nhỏ hơn sẽ in ra trước

```python
s=input()
a=list(s.split())
a.sort(key=lambda x:(x[0],len(x)))
print(*a)
```

Tìm xâu con chứa các ký tự giống nhau liên tiếp dài nhất, nếu có nhiều xâu thỏa mãn thì chọn xâu có thứ tự từ điển lớn nhất (chuỗi in thường)

```python
s=input()
d={}
dem=1
for i in range(1,len(s)):
    if s[i]==s[i-1]:
        dem+=1
        d[s[i-1]]=1*dem
    else:
        dem=1
d1=dict(sorted(d.items(),reverse=True))
max_val=0
for val in d1.values():
    max_val=max(max_val,val)
for key,val in d1.items():
    if val==max_val:
        print("".join([key]*val))
        break
```
