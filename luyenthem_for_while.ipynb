```python
a,b,n=map(int,input().split())
for x in range(0,n//a+1): # Vì khi y=0 thì ax=n (khi này b max=n/a)
    if (n-a*x)%b==0:
        print(True)
        break
else:
    print(False)
```

Nhập vào một số nguyên dương N. Hãy tìm và in ra N số nguyên dương đầu tiên thỏa mãn điều kiện: Tổng các chữ số của nó là một số nguyên tố.
(Ví dụ: Số 14 có tổng chữ số là 1+4 = 5. Vì 5 là số nguyên tố nên 14 thỏa mãn).

```python
import math
def prime(n):
    if n<2:
        return False
    for i in range(2,int(math.sqrt(n)+1)):
        if n%i==0:
            return False
    return True

def shp(a):
    tong=0
    while a>0:
        tong+=a%10
        a=a//10
    if prime(tong):
        return True
    return False

b=int(input())
dem=0
so=1
while dem<b:
    if shp(so):    
        print(so,end=" ")
        dem+=1
    so+=1
```

Nhập vào một số nguyên dương n. Áp dụng quy tắc sau: Nếu n chẵn, chia n cho 2. Nếu $X$ lẻ, nhân n với 3 rồi cộng 1. Lặp lại quá trình này cho đến khi n = 1.

```python
n=int(input())
buoc=0
max_val=0
while n!=1:
    if n%2==0:
        n=n//2
        buoc+=1
        max_val=max(max_val,n)
    else:
        n=3*n+1
        buoc+=1
        max_val=max(max_val,n)
print(buoc, max_val)
    
```

Nhập vào một số nguyên dương n. Tìm và in ra tất cả các bộ ba số nguyên dương a, b, c sao cho a^2 + b^2 = c^2, với c<=n, và a, b, c đôi một nguyên tố cùng nhau (Ước chung lớn nhất của chúng bằng 1).

```python
def ucln(a, b):
    while b > 0:
        temp = b
        b = a % b
        a = temp
    return a

n = int(input("Nhập số nguyên dương n: "))
for a in range(1, n + 1):
    for b in range(a + 1, n + 1):
        c_square = a**2 + b**2
        c = int(c_square ** 0.5)
        
        # Kiểm tra 3 điều kiện:
        # 1. c <= n (theo yêu cầu đề bài)
        # 2. c_square phải là số chính phương (c**2 == c_square)
        # 3. a và b nguyên tố cùng nhau (ucln == 1)
        if c <= n and c**2 == c_square and ucln(a, b) == 1:
            print(f"({a}, {b}, {c})")

    
```
