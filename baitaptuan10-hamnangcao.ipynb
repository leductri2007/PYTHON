W9A1
Viết hàm merge hai danh sách đã sắp xếp mà không dùng bất cứ
thuật toán sắp xếp nào

```python
def merge(a,b):
    i=0
    j=0
    c=[]
    while i<len(a) and j<len(b):
        if a[i]<b[j]:
            c.append(a[i])
            i+=1
        elif a[i]>b[j]:
            c.append(b[j])
            j+=1
        else:
            c.append(a[i])
            i+=1
            
    while i<len(a):
        c.append(a[i])
        i+=1
    while j<len(b):
        c.append(b[j])
        j+=1
    return c
a=list(map(int,input().split()))
b=list(map(int,input().split()))
print(merge(a,b))
    
```

W9A3
Cho một số nguyên dương, hãy thực hiện hoán đổi đúng 2 chữ
số để được một số nguyên nhỏ nhất

```python
n=int(input())
b=list(str(n))
a=[]
c=[]
for x in b:
    a.append(int(x))
    if x!=0:
        c.append(int(x))
min1=min(c)
c.remove(min1)
min2=min(c)
a.insert(min2,0)
a.insert(min1,0)
k="".join(a)
print(k)
```
