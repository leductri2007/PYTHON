Bubble sort

```python
def bubble_sort_basic(a): # duyệt lần lượt
    for i in range(len(a)):
        # Vòng lặp trong thực hiện so sánh các cặp kề nhau
        for j in range(len(a) - i - 1): # -i : bỏ qua các phần tử đã ổn định
            if a[j] > a[j + 1]:
                # Hoán đổi nếu phần tử trước lớn hơn phần tử sau
                a[j], a[j + 1] = a[j + 1], a[j]
    return a
```

Insertion sort

```python
def insertion_sort(a):
    # Vòng lặp ngoài: Chạy từ phần tử thứ 2 (chỉ số 1) đến hết mảng
    for i in range(1, len(a)):
        key = a[i]  # "Lá bài mới" cần chèn
        j = i - 1     # Vị trí của phần tử cuối cùng trong phần đã sắp xếp
        
        # Vòng lặp trong: Di chuyển các phần tử của phần đã sắp xếp 
        # lớn hơn 'key' sang bên phải một vị trí
        while j >= 0 and a[j] > key:
            a[j + 1] = a[j]  # Dịch chuyển sang phải
            j -= 1               # Lùi về phía trước để kiểm tra tiếp
            
        # Chèn 'key' vào vị trí thích hợp vừa lộ ra
        a[j + 1] = key
    return a
```

Merge sort

```python
# TRỘN 2 MẢNG
def merge(left, right):
    result = []
    i = j = 0  # i là con trỏ mảng left, j là con trỏ mảng right
    
    # So sánh từng phần tử đầu tiên của hai mảng con
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1  # Dịch con trỏ mảng trái sang phải
        else:
            result.append(right[j])
            j += 1  # Dịch con trỏ mảng phải sang phải
            
    # Nếu 1 TRONG 2 mảng con vẫn còn phần tử THỪA thì thêm vào cuối
    result.extend(left[i:])
    result.extend(right[j:])
    return result

def merge_sort(a):
    # ĐIỀU KIỆN DỪNG ĐỆ QUY: Nếu mảng chỉ còn 1 phần tử hoặc rỗng thì không cần chia nữa
    if len(a) <= 1:
        return a
        
    # BƯỚC 1: Tìm vị trí chính giữa để chia đôi mảng
    mid = len(a) // 2
    left_half = a[:mid]   # Mảng con bên trái
    right_half = a[mid:]  # Mảng con bên phải
    
    # BƯỚC 2: chia đệ quy hai nửa trái và phải
    left_sorted = merge_sort(left_half)  # mang nửa bên trái này đi cắt đôi tiếp, cắt cho đến khi nào nó tự sắp xếp xong
    right_sorted = merge_sort(right_half)
    
    # BƯỚC 3: Trộn hai nửa đã sắp xếp lại với nhau
    return merge(left_sorted, right_sorted) # merge chỉ trộn 2 mảng đ

```
