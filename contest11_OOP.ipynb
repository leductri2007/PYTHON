Viết chương trình khai báo lớp phân số, gồm 2 thuộc tính là tử và mẫu, nhận vào một phân số và in ra dạng tối giản của phân số đó

```python
import math
class phan_so:
    def __init__(self,tu_so,mau_so):
        self.tu_so=tu_so
        self.mau_so=mau_so
    def toi_gian(self):
        ucln=math.gcd(self.tu_so,self.mau_so)
        self.tu_so//=ucln
        self.mau_so//=ucln
    def __str__(self):
        return f"{self.tu_so}/{self.mau_so}"

tu_so,mau_so=map(int,input().split())
p=phan_so(tu_so,mau_so)
p.toi_gian()
print(p)
```

Viết chương trình khai báo lớp Thí Sinh gồm các thông tin: Họ tên, ngày sinh, điểm môn 1, điểm môn 2, tổng điểm. Đọc thông tin 1 thí sinh từ bàn phím và in ra màn hình thông tin: Họ tên, ngày sinh, tổng điểm

```python
class Thi_sinh():
    def __init__(self,ten,ngay_sinh,diem1,diem2):
        self.ten=ten
        self.ngay_sinh=ngay_sinh
        self.diem1=diem1
        self.diem2=diem2
    def __str__(self):
        tong_diem=diem1+diem2
        return f"{self.ten} {self.ngay_sinh} {tong_diem:.1f}"
ten=input()
ngay_sinh=input()
diem1,diem2=map(int,input().split())
t=Thi_sinh(ten,ngay_sinh,diem1,diem2)
print(t)
```

Viết chương trình khai báo lớp Sinh Viên gồm các thông tin: Mã SV, Họ tên, Lớp, Ngày sinh và Điểm GPA (dạng số thực float). Hàm khởi tạo không có tham số, gán các giá trị thuộc tính ở trạng thái mặc định (xâu ký tự rỗng, giá trị số bằng 0). Đọc thông tin 1 sinh viên từ bàn phím (không có mã sinh viên) và in ra màn hình. Trong đó Mã SV được gán là SV001. Ngày sinh được chuẩn hóa về dạng dd/mm/yyyy.

```python
class sv:
    def __init__(self):
        self.msv=""
        self.ten=""
        self.lop=""
        self.ngay_sinh=""
        self.gpa=""
    def chuan_hoa(self):
        if self.ngay_sinh[1]=="/":
            self.ngay_sinh="0"+self.ngay_sinh
        elif self.ngay_sinh[4]=="/":
            self.ngay_sinh=self.ngay_sinh[0:3]+"0"+self.ngay_sinh[3:]
    def __str__(self):
        return f"{self.msv},{self.ten},{self.lop},{self.ngay_sinh},{self.gpa:.1f}"
ten=input()
lop=input()
ngay_sinh=input()
gpa=float(input())
s=sv()
s.msv="SV001"
s.ten=ten
s.lop=lop
s.gpa=gpa
s.chuan_hoa()
print(s)
```

Trường cần quản lý các đối tượng là sinh viên và giáo viên. Sinh viên gồm các thông tin : mã sinh viên, tên, ngày sinh, địa chỉ, lớp, điểm gpa. Giáo viên gồm các thông tin : mã giáo viên, tên, ngày sinh, địa chỉ, khoa, lương, lớp mà giáo viên này phụ trách. Thực hiện đọc các thông tin danh sách sinh viên và giáo viên từ bàn phím sau đó chuẩn hóa tên, ngày sinh, biết rằng sinh viên sẽ có mã bắt đầu bằng SV (ví dụ SV112), giáo viên có mã bắt đầu bằng GV (ví dụ GV222) sau đó hiển thị giáo viên phụ trách và các sinh viên thuộc về 1 lớp theo truy vấn.  

Input Format  

Dòng đầu tiên là N : số lượng giáo viên và sinh viên. Các dòng tiếp theo mô tả thông tin của giáo viên hoặc sinh viên, mỗi thông tin gồm 6 dòng, đối với sinh viên 6 dòng gồm : mã sinh viên, tên, ngày sinh, địa chỉ, lớp, điểm gpa, đối với giáo viên 7 dòng gồm : mã giáo viên, tên, ngày sinh, địa chỉ, khoa, lương, lớp phụ trách. Dòng cuối cùng trong input là tên lớp cần truy vấn.

Output Format  

Đầu tiên in ra giáo viên phụ trách lớp, mỗi giáo viên in ra thông tin trên 1 dòng, các thông tin cách nhau một dấu cách, một lớp có thể có nhiều giáo viên cùng phụ trách. Khi đó hãy liệt kê giáo viên theo danh sách. Những dòng tiếp theo in ra danh sách sinh viên, mỗi sinh viên in ra thông tin trên 1 dòng, các thông tin cách nhau một dấu cách, gpa in 2 số sau dấu phẩy theo danh sách.
