1. Khởi tạo với danh sách rỗng
2. Thêm vào giỏ hàng một sản phẩm mới
3. Xóa khỏi giỏ một sản phẩm nào đó (tìm theo tên để xóa)
4. Kiểm tra xem giỏ hàng có rỗng không
5. Tính tổng tiền giỏ hàng
6. Hiển thị toàn bộ sản phẩm trong giỏ

Chương trình dừng khi người dùng nhập "yes" cho câu hỏi "Có
muốn thoát khỏi chương trình không?" 

```python
class ShoppingCart:
    def __init__(self):
        self.ds=[]
    def them(self,name,price):
        ds0={"name":name,"price":price}
        self.ds.append(ds0)
    def xoa(self,name):
        tim_thay=False
        for ds0 in self.ds:
            if ds0["name"].lower()==name.lower():
                self.ds.remove(ds0)
                tim_thay=True
                break
    def check(self):
        if len(self.ds)==0:
            return True
        return False
    def tong_tien(self):
        tong=0
        for ds0 in self.ds:
            tong+=ds0["price"]
        return tong
    def hien_thi(self):
        return self.ds
    def xoa_toan_bo(self):
        self.ds.clear()

a=ShoppingCart()
while True:
    lua_chon=int(input())
    if lua_chon==2:
        name=input()
        price=float(input())
        a.them(name,price)
    elif lua_chon==3:
        name=input()
        a.xoa(name)
    elif lua_chon==4:
        print(a.check())
    elif lua_chon==5:
        print(a.tong_tien())
    elif lua_chon==6:
        print(a.hien_thi())
    elif lua_chon==7:
        a.xoa_toan_bo()
    print("Bạn có muốn thoát k?")
    k=input()
    if k=="yes":
        break

        
```

Yêu cầu
Lớp Fraction(num, den) (tử, mẫu nguyên, mẫu ≠ 0) với:
Rút gọn phân số trong __init__ (quy ước mẫu > 0).
Các phương thức add/sub/mul/div trả về phân số đã rút gọn.
__str__() in dạng a/b.
Input
a b op c d (tương ứng a/b (op) c/d, op thuộc {+,-,*,/})
Output
Phân số kết quả đã rút gọn 

```python
import math
class Fraction:
    def __init__(self,tu,mau):
        ucln=math.gcd(tu,mau)
        self.tu=tu//ucln
        self.mau=mau//ucln
    def cong(self,ps):
        tu_moi=self.tu*ps.mau+ps.tu*self.mau
        mau_moi=self.mau*ps.mau
        return Fraction(tu_moi,mau_moi)
    def tru(self,ps):
        tu_moi=self.tu*ps.mau-ps.tu*self.mau
        mau_moi=self.mau*ps.mau
        return Fraction(tu_moi,mau_moi)
    def nhan(self,ps):
        tu_moi=self.tu*ps.tu
        mau_moi=self.mau*ps.mau
        return Fraction(tu_moi,mau_moi)
    def chia(self,ps):
        tu_moi=self.tu*ps.mau
        mau_moi=self.mau*ps.tu
        return Fraction(tu_moi,mau_moi)
    def __str__(self):
        return f"{self.tu}/{self.mau}"
a,b=map(int,input().split())
op=input()
c,d=map(int,input().split())
ps1=Fraction(a,b)
ps2=Fraction(c,d)
if op=="+":
    print(ps1.cong(ps2))
elif op=="-":
    print(ps1.tru(ps2))
elif op=="*":
    print(ps1.nhan(ps2))
elif op==":":
    print(ps1.chia(ps2))
    
```

Lớp Student(name) với:
add_score(subject, score)
avg() trả về điểm trung bình.
rank() trả về "Excellent" nếu avg ≥ 8, "Good" nếu 6.5–<8,
"Average" nếu 5–<6.5, ngược lại "Poor".
Input
Dòng 1: name
Dòng 2: n
n dòng tiếp: subject score (score là số thực 0–10)
Output
name avg rank (avg 2 chữ số). 

```python
class Student:
    def __init__(self,name):
        self.name=name
        self.ds=[]
    def add_score(self,diem):
        self.ds.append(diem)
    def avg(self):
        tb=sum(self.ds)/len(self.ds)
        return tb
    def rank(self):
        if self.avg()>=8:     ###self.avg()
            return "Excellent"
        elif 6.5<=self.avg()<8:
            return "Good"
        elif 5<=self.avg()<6.5:
            return "Average"
        elif self.avg()<5:
            return "Poor"
name=input()
a=Student(name)
n=int(input())
for _ in range(n):
    dau_vao=input().split()
    diem=float(dau_vao[1])
    a.add_score(diem)
print(f"{name} {a.avg()} {a.rank()}")
```

Viết hàm khởi tạo ComplexNumber(double initRealPart, double
initImaginaryPart) để khởi tạo các giá trị phần thực và phần ảo

multiply(complexNumber other) trả về tích của số phức hiện tại
và một số phức khác.
Đầu vào
Đầu vào từ bàn phím gồm hai dòng, mỗi dòng gồm hai số thực
cách nhau bởi một dấu cách biểu diễn phần thực và phần ảo
của một số phức. Các số trong đầu vào có giá trị tuyệt đối
không vượt quá \(100\).
Đầu ra
In ra màn hình hai số biểu diễn phần thực và phần ảo của tích
hai số phức

```python
class Complexnumber:
    def __init__(self,thuc,ao):
        self.thuc=thuc
        self.ao=ao
    def tich(self,other):
        phan_thuc=self.thuc*other.thuc-self.ao*other.ao
        phan_ao=self.thuc*other.ao+self.ao*other.thuc
        return f"{phan_thuc} {phan_ao}"
a,b=map(int,input().split())
z1=Complexnumber(a,b)
c,d=map(int,input().split())
z2=Complexnumber(c,d)
k=z1.tich(z2)
print(k)
```

[Triangle]
Khai báo cấu trúc dữ liệu tam giác Triangle chứa 3 biến
(egde1,edge2,edge3) kiểu int là độ dài 3 cạnh của
tam giác đó.
Viết hàm khởi tạo Triangle(int initEdge1,int initEdge2, int
initEdge3) để khởi tạo độ dài các cạnh của tam giác và hàm
double getArea() trả về diện tích của tam giác đó (nếu tam giác
không hợp lệ, trả về -1).
Hoàn thiện chương trình dưới đây nhập vào 3 cạnh của một
tam giác và in ra diện tích của tam giác đó.
Đầu vào
Một dòng duy nhất từ bàn phím chứa 3 số nguyên là độ dài
3 cạnh của tam giác
Đầu ra
In ra màn hình diện tích tam giác (làm tròn đến 2 chữ số thập
phân sau dấu phẩy). In ra invalid nếu 3 số nhập vào
không phải là 3 cạnh của một tam giác 

```python
import math
class Triangle:
    def __init__(self,c1,c2,c3):
        self.c1=c1
        self.c2=c2
        self.c3=c3
    def getArea(self):
        if self.c1+self.c2>self.c3 and self.c1+self.c3>self.c2 and self.c2+self.c3>self.c1:
            p=(self.c1+self.c2+self.c3)/2
            s=math.sqrt((p*(p-self.c1)*(p-self.c2)*(p-self.c3)))
            return f"{s:.2f}"
        else:
            return -1
c1,c2,c3=map(int,input().split())
a=Triangle(c1,c2,c3)

if a.getArea()==-1:
    print("invalid")
else:
    print(a.getArea())
```
