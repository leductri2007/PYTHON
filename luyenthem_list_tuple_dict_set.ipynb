Bạn được cho 3 số nguyên x,y,z đại diện cho kích thước của một hình hộp chữ nhật và một số nguyên n.

Nhiệm vụ: Hãy tạo một danh sách tất cả các tọa độ (i,j,k) có thể có trên lưới 3D sao cho:

0≤i≤x

0≤j≤y

0≤k≤z

Điều kiện quan trọng: Tổng của i+j+k phải khác n

```python
x,y,z,n=map(int,input().split())
for i in range(x+1):
    for j in range(y+1):
        for k in range(z+1):
            if i+j+k!=n:
                print([i,j,k])
```

Cho tên và điểm số của n học sinh trong một lớp. Hãy lưu trữ chúng trong một danh sách lồng nhau (nested list) và in ra tên của những học sinh có điểm thấp thứ nhì.Lưu ý:Nếu có nhiều học sinh cùng có mức điểm thấp thứ nhì, hãy sắp xếp tên của họ theo thứ tự bảng chữ cái (alphabet) và in mỗi tên trên một dòng mới.

```python
n=int(input())
a=[]
for i in range(n):
    name=input()
    score=float(input())
    a.append([name,score])
a.sort(key=lambda x:x[0])
c=[]
for name,score in a:
    c.append(score)
d=list(set(c))
d.sort()
for name,score in a:
    if score==d[1]:
        print(name)

```

Hệ thống sẽ đọc vào một danh sách sinh viên, mỗi sinh viên đi kèm với một danh sách các điểm số. Dữ liệu được lưu dưới dạng dictionary với cặp tên_sinh_viên: [danh_sách_điểm].

Nhiệm vụ:
Sau khi nhập hết dữ liệu, bạn sẽ nhận được một cái tên sinh viên cụ thể (query_name). Hãy tính điểm trung bình của sinh viên đó và in ra màn hình với chính xác 2 chữ số sau dấu phẩy.

```python
n=int(input())
c=[]
for _ in range(n):
    name=input()
    score=list(map(float,(input().split())))
    c.append([name,score])
d=dict(c)
query_name=input()
for key,val in d.items():
    if key==query_name:
        print(sum(val)/len(val))
        break
    
```
