Yêu cầu: Cho một list các chuỗi strs = ["eat", "tea", "tan", "ate", "nat", "bat"]. Hãy viết hàm trả về một danh sách các nhóm từ có cùng tập hợp ký tự.

```python
def check(strs):
    d={}
    for x in strs:
        x1=tuple(sorted(x))
        if x1 not in d:
            d[x1]=[x]
        else:
            d[x1].append(x)
    return list(d.values())
strs=["eat", "tea", "tan", "ate", "nat", "bat","eta"]
print(check(strs))
```

4. Nén Chuỗi (String Compression)
Yêu cầu: Viết hàm nén chuỗi theo quy tắc: "aaabbccccd" thành "a3b2c4d1". Nếu chuỗi sau khi nén không ngắn hơn chuỗi gốc, hãy trả về chuỗi gốc.

```python
def nen(s):
    d=""
    dem=1
    for i in range(1,len(s)):
        
        if s[i]==s[i-1]:
            dem+=1
        else:
            d=d+s[i-1]+str(dem)
            dem=1
    d=d+s[-1]+str(dem) #cộng chữ cuối cùng
    return d
s=input()
print(nen(s))
    
```

. Tìm phần tử lặp lại đầu tiên (First Recurring Character)
Yêu cầu: Cho một chuỗi rất dài, tìm ký tự đầu tiên xuất hiện lần thứ hai.

Ví dụ: "ABCA" -> Trả về "A". "BCABA" -> Trả về "B".

Yêu cầu phụ: Tối ưu bộ nhớ nếu tập ký tự chỉ là mã ASCII.

```python
def first_recurring(s):
    seen = set()
    for char in s:
        if char not in seen:
            seen.add(char)
        else:
            return char
    
s=input()
print(first_recurring(s))

```
