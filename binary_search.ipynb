Thuật toán binary search

```python
def binary(a,l,r,x):
    while l<=r:
        m=(l+r)//2
        if a[m]==x:
            return True
        elif a[m]<x:
            l=m+1
        else:
            r=m-1
a=list(map(int,(input().split())))
print(binary(a,0,len(a)-1,5))
```

Tìm vị trí xuất hiện đầu tiên của phần tử có giá trị x trong list đã sắp tăng dần (nếu x xuất hiện nhiều lần, tìm vị trí đầu tiên)

```python
def binary_first(a,l,r,x):
    kq=-1
    while l<=r:
        m=(l+r)//2
        if a[m]==x:
            kq=m
            r=m-1 # vì tìm vị trí đầu tiên nên tìm tiếp bên trái
        elif a[m]<x:
            l=m+1
        else:
            r=m-1
    return kq
a=list(map(int,(input().split())))
print(binary_first(a,0,len(a)-1,5)) # r=len(a)-1
```

Tìm vị trí xuất hiện đầu tiên của phần tử có giá trị >=x trong list đã sắp tăng dần 

```python
def binary_first(a,l,r,x):
    kq=-1
    while l<=r:
        m=(l+r)//2
        if a[m]>=x:
            kq=m
            r=m-1 
        else:
            l=m+1
    return kq
a=list(map(int,(input().split())))
print(binary_first(a,0,len(a)-1,5)) # r=len(a)-1
```
