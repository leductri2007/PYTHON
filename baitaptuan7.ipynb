Viết hàm tìm kiếm nhị phân
trên một danh sách đã
được sắp xếp tăng dần.
Trả về chỉ số của phần tử
hoặc -1 nếu không tìm
thấy. mẫu hàm: def
binary_search(arr, left,
right, target):

```python
def binary_search(a,l,r,x):
    
    while l<=r:
        m=(l+r)//2
        if a[m]==x:
            return m
        if a[m]<x:
            l=m+1
        elif a[m]>x:
            r=m-1
    return -1
a=list(map(int,input().split()))
x=int(input())
print(binary_search(a,0,len(a)-1,x))
```

W7A3
Viết hàm thực hiện thuật
toán bubble sort, Selection
Sort để sắp xếp danh sách
theo thứ tự tăng dần. Sắp
xếp nổi bọt (Bubble Sort);Sắp xếp chọn (Selection
Sort)

```python
def bubble_sort(a):
    n = len(a)
    for i in range(n):
        # n-i-1 vì các phần tử cuối đã nằm đúng vị trí rồi
        for j in range(0, n - i - 1):
            if a[j] > a[j + 1]:
                # Đổi chỗ (Swap)
                a[j], a[j + 1] = a[j + 1], a[j]
    return a

def selection_sort(a):
    n = len(a)
    for i in range(n):
        min_idx = i # Giả định vị trí đầu tiên là nhỏ nhất
        for j in range(i + 1, n):
            if a[j] < a[min_idx]:
                min_idx = j # Cập nhật vị trí của số nhỏ hơn
        
        # Đổi chỗ số nhỏ nhất tìm được với vị trí i
        a[i], a[min_idx] = a[min_idx], a[i]
    return a
```

Viết hàm tìm phần tử có số
lần xuất hiện nhiều nhất
trong danh sách. Nếu có
nhiều phần tử cùng số lần,
trả về phần tử xuất hiện
trước.

```python
def find(a):
    d={}
    for x in a:
        d[x]=d.get(x,0)+1
    max_val=0
    for value in d.values():
        max_val=max(max_val,value)
    for key, value in d.items():
        if value==max_val:
            return key
a=list(map(int,input().split()))
print(find(a))
```

W7A7
Cho một danh sách các số
nguyên nums, hãy tìm độ
dài của dãy con tăng
nghiêm ngặt dài nhất
(không nhất thiết phải liền kề).

```python
def do_dai(a):
    b=[]
    for x in a:
        l=0
        r=len(b)
        while l<r:
            m=(l+r)//2
            if b[m]<x:
                l=m+1
            else:
                r=m

        if l==len(b):
            b.append(x)
        else:
            b[l]=x
    return len(b),b
a=list(map(int,input().split()))
print(do_dai(a))

        
```

Dãy hài hòa là một dãy
thỏa mãn điều kiện giá trị
lớn nhất và giá trị nhỏ nhất
của dãy khác nhau đúng
bằng 1.
Viết hàm trả về độ dài của
dãy hài hòa con dài nhất có
thể có trong mảng giá trị
nguyên arr có n
phần tử
Ví dụ:
Input: 1 3 2 2 5 2 3 7
Output: 5 - Dãy hài hòa con
dài nhất là [3,2,2,2,3]


```python
from collections import Counter

def dhh(a):
    cnt=Counter(a)
    max_len=0
    for x in cnt:
        if x+1 in cnt:
            do_dai=cnt[x]+cnt[x+1]
            max_len=max(max_len,do_dai)
    return max_len
a=list(map(int,input().split()))
print(dhh(a))
```
